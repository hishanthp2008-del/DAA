```python
# ---------------------------------------------------------
# Implementation of Kruskal's and Prim's Algorithms
# for Minimum Spanning Tree (MST)
#
# Graph:
# Vertices = 0, 1, 2, 3, 4, 5, 6
# Edges = 9
# ---------------------------------------------------------


# ---------------------------------------------------------
# GRAPH DEFINITION
# ---------------------------------------------------------

vertices = 7

# (u, v, weight)
edges = [
    (0, 1, 4),
    (0, 2, 3),
    (1, 2, 1),
    (1, 3, 2),
    (2, 3, 4),
    (2, 4, 5),
    (3, 4, 2),
    (3, 5, 3),
    (4, 6, 3)
]


# ---------------------------------------------------------
# KRUSKAL'S ALGORITHM
# ---------------------------------------------------------

class DisjointSet:
    def __init__(self, n):
        self.parent = list(range(n))
        self.rank = [0] * n

    def find(self, x):
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])
        return self.parent[x]

    def union(self, x, y):
        root_x = self.find(x)
        root_y = self.find(y)

        if root_x == root_y:
            return False

        # Union by rank
        if self.rank[root_x] < self.rank[root_y]:
            self.parent[root_x] = root_y

        elif self.rank[root_x] > self.rank[root_y]:
            self.parent[root_y] = root_x

        else:
            self.parent[root_y] = root_x
            self.rank[root_x] += 1

        return True


def kruskal(vertices, edges):

    # Sort edges according to weight
    sorted_edges = sorted(edges, key=lambda edge: edge[2])

    ds = DisjointSet(vertices)

    mst = []
    total_weight = 0

    for u, v, weight in sorted_edges:

        # Add edge if it does not form a cycle
        if ds.union(u, v):
            mst.append((u, v, weight))
            total_weight += weight

        # MST contains V-1 edges
        if len(mst) == vertices - 1:
            break

    return mst, total_weight


# ---------------------------------------------------------
# PRIM'S ALGORITHM
# ---------------------------------------------------------

def prim(vertices, edges):

    # Create adjacency list
    graph = [[] for _ in range(vertices)]

    for u, v, weight in edges:
        graph[u].append((v, weight))
        graph[v].append((u, weight))

    # Start from vertex 0
    visited = [False] * vertices
    mst = []
    total_weight = 0

    # Simple list used as a priority queue
    priority_queue = [(0, 0, -1)]  # (weight, vertex, parent)

    while priority_queue and len(mst) < vertices - 1:

        # Find minimum-weight edge
        min_index = 0

        for i in range(1, len(priority_queue)):
            if priority_queue[i][0] < priority_queue[min_index][0]:
                min_index = i

        weight, vertex, parent = priority_queue.pop(min_index)

        # Ignore already visited vertices
        if visited[vertex]:
            continue

        visited[vertex] = True

        # Add edge to MST (except starting vertex)
        if parent != -1:
            mst.append((parent, vertex, weight))
            total_weight += weight

        # Add adjacent edges
        for neighbor, edge_weight in graph[vertex]:

            if not visited[neighbor]:
                priority_queue.append(
                    (edge_weight, neighbor, vertex)
                )

    return mst, total_weight


# ---------------------------------------------------------
# PRINT MST
# ---------------------------------------------------------

def print_mst(name, mst, total_weight):

    print("\n" + "=" * 50)
    print(name)
    print("=" * 50)

    print(f"{'Edge':<15}{'Weight':>10}")
    print("-" * 25)

    for u, v, weight in mst:
        print(f"{u} - {v:<10}{weight:>5}")

    print("-" * 25)
    print(f"Total MST Weight: {total_weight}")


# ---------------------------------------------------------
# MAIN PROGRAM
# ---------------------------------------------------------

def main():

    print("=" * 50)
    print("MINIMUM SPANNING TREE")
    print("=" * 50)

    print("\nGraph edges:")
    for u, v, weight in edges:
        print(f"{u} -- {v}  (weight = {weight})")

    # Run Kruskal
    kruskal_mst, kruskal_weight = kruskal(
        vertices, edges
    )

    # Run Prim
    prim_mst, prim_weight = prim(
        vertices, edges
    )

    # Display results
    print_mst(
        "Kruskal's Algorithm",
        kruskal_mst,
        kruskal_weight
    )

    print_mst(
        "Prim's Algorithm",
        prim_mst,
        prim_weight
    )

    # -----------------------------------------------------
    # VERIFY RESULTS
    # -----------------------------------------------------

    print("\n" + "=" * 50)
    print("VERIFICATION")
    print("=" * 50)

    print("Kruskal MST weight :", kruskal_weight)
    print("Prim MST weight    :", prim_weight)

    if kruskal_weight == prim_weight:
        print("\n✓ Both algorithms produce the same MST weight.")
        print(f"✓ Minimum total weight = {kruskal_weight}")
    else:
        print("\n✗ MST weights are different.")

    # Check number of edges
    print(f"\nKruskal MST edges: {len(kruskal_mst)}")
    print(f"Prim MST edges   : {len(prim_mst)}")

    if len(kruskal_mst) == vertices - 1:
        print("✓ Kruskal produces V-1 edges.")

    if len(prim_mst) == vertices - 1:
        print("✓ Prim produces V-1 edges.")


# ---------------------------------------------------------
# EXECUTE
# ---------------------------------------------------------

if __name__ == "__main__":
    main()
```

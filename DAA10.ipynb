```python
# ---------------------------------------------------------
# QUICK SORT PERFORMANCE COMPARISON
# Deterministic Quick Sort vs Randomized Quick Sort
# ---------------------------------------------------------

import random
import time
import sys

# Increase recursion limit for worst-case Quick Sort
sys.setrecursionlimit(30000)


# ---------------------------------------------------------
# PARTITION FUNCTION
# ---------------------------------------------------------

def partition(arr, low, high, pivot_index, counter):
    """
    Lomuto partition.
    pivot_index can be the last element or a random element.
    """

    # Move selected pivot to the end
    arr[pivot_index], arr[high] = arr[high], arr[pivot_index]

    pivot = arr[high]

    i = low - 1

    for j in range(low, high):

        # Count comparison between arr[j] and pivot
        counter[0] += 1

        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]

    # Put pivot in its correct position
    arr[i + 1], arr[high] = arr[high], arr[i + 1]

    return i + 1


# ---------------------------------------------------------
# 1. DETERMINISTIC QUICK SORT
# Pivot = Last Element
# ---------------------------------------------------------

def deterministic_quick_sort(arr, low, high, counter):

    if low < high:

        # Last element is used as pivot
        pivot_index = high

        p = partition(
            arr,
            low,
            high,
            pivot_index,
            counter
        )

        # Sort left part
        deterministic_quick_sort(
            arr,
            low,
            p - 1,
            counter
        )

        # Sort right part
        deterministic_quick_sort(
            arr,
            p + 1,
            high,
            counter
        )


# ---------------------------------------------------------
# 2. RANDOMIZED QUICK SORT
# Pivot = Random Element
# ---------------------------------------------------------

def randomized_quick_sort(arr, low, high, counter):

    if low < high:

        # Select random pivot
        pivot_index = random.randint(low, high)

        p = partition(
            arr,
            low,
            high,
            pivot_index,
            counter
        )

        # Sort left part
        randomized_quick_sort(
            arr,
            low,
            p - 1,
            counter
        )

        # Sort right part
        randomized_quick_sort(
            arr,
            p + 1,
            high,
            counter
        )


# ---------------------------------------------------------
# GENERATE TEST ARRAYS
# ---------------------------------------------------------

def generate_array(size, configuration):

    if configuration == "Random":
        return [
            random.randint(1, 100000)
            for _ in range(size)
        ]

    elif configuration == "Sorted":
        return list(range(size))

    elif configuration == "Reverse Sorted":
        return list(range(size, 0, -1))

    elif configuration == "Nearly Sorted":

        arr = list(range(size))

        # Perform a small number of random swaps
        swaps = max(1, size // 100)

        for _ in range(swaps):
            i = random.randint(0, size - 1)
            j = random.randint(0, size - 1)

            arr[i], arr[j] = arr[j], arr[i]

        return arr

    else:
        raise ValueError("Unknown configuration")


# ---------------------------------------------------------
# MEASURE DETERMINISTIC QUICK SORT
# ---------------------------------------------------------

def test_deterministic(arr):

    counter = [0]

    start_time = time.perf_counter()

    deterministic_quick_sort(
        arr,
        0,
        len(arr) - 1,
        counter
    )

    end_time = time.perf_counter()

    execution_time = end_time - start_time

    return execution_time, counter[0]


# ---------------------------------------------------------
# MEASURE RANDOMIZED QUICK SORT
# ---------------------------------------------------------

def test_randomized(arr):

    counter = [0]

    start_time = time.perf_counter()

    randomized_quick_sort(
        arr,
        0,
        len(arr) - 1,
        counter
    )

    end_time = time.perf_counter()

    execution_time = end_time - start_time

    return execution_time, counter[0]


# ---------------------------------------------------------
# MAIN PROGRAM
# ---------------------------------------------------------

def main():

    SIZE = 10000

    configurations = [
        "Random",
        "Sorted",
        "Reverse Sorted",
        "Nearly Sorted"
    ]

    print("=" * 95)
    print("DETERMINISTIC VS RANDOMIZED QUICK SORT")
    print("=" * 95)

    print(f"\nArray Size: {SIZE}")

    print("\nPivot Selection:")
    print("Deterministic Quick Sort -> Last Element")
    print("Randomized Quick Sort    -> Random Element")

    print("\n" + "-" * 95)

    print(
        f"{'Configuration':<20}"
        f"{'Algorithm':<25}"
        f"{'Time (sec)':<15}"
        f"{'Comparisons':<20}"
    )

    print("-" * 95)

    results = []

    for configuration in configurations:

        # Generate original array
        original = generate_array(
            SIZE,
            configuration
        )

        # -------------------------------------------------
        # Deterministic Quick Sort
        # -------------------------------------------------

        arr1 = original.copy()

        deterministic_time, deterministic_comparisons = (
            test_deterministic(arr1)
        )

        # -------------------------------------------------
        # Randomized Quick Sort
        # -------------------------------------------------

        arr2 = original.copy()

        randomized_time, randomized_comparisons = (
            test_randomized(arr2)
        )

        # Verify that both arrays are sorted
        assert arr1 == arr2
        assert arr1 == sorted(original)

        # Store results
        results.append(
            (
                configuration,
                deterministic_time,
                deterministic_comparisons,
                randomized_time,
                randomized_comparisons
            )
        )

        # Print results
        print(
            f"{configuration:<20}"
            f"{'Deterministic':<25}"
            f"{deterministic_time:<15.6f}"
            f"{deterministic_comparisons:<20}"
        )

        print(
            f"{'':<20}"
            f"{'Randomized':<25}"
            f"{randomized_time:<15.6f}"
            f"{randomized_comparisons:<20}"
        )

        print("-" * 95)

    # -----------------------------------------------------
    # SUMMARY
    # -----------------------------------------------------

    print("\n" + "=" * 95)
    print("SUMMARY")
    print("=" * 95)

    for (
        configuration,
        deterministic_time,
        deterministic_comparisons,
        randomized_time,
        randomized_comparisons
    ) in results:

        print(f"\n{configuration}:")

        print(
            f"  Deterministic -> "
            f"{deterministic_time:.6f} sec, "
            f"{deterministic_comparisons} comparisons"
        )

        print(
            f"  Randomized    -> "
            f"{randomized_time:.6f} sec, "
            f"{randomized_comparisons} comparisons"
        )

        if deterministic_comparisons < randomized_comparisons:
            print("  Fewer comparisons: Deterministic")

        elif randomized_comparisons < deterministic_comparisons:
            print("  Fewer comparisons: Randomized")

        else:
            print("  Comparisons: Equal")


# ---------------------------------------------------------
# EXECUTE PROGRAM
# ---------------------------------------------------------

if __name__ == "__main__":
    main()
```

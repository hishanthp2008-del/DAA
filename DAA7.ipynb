```python
# ---------------------------------------------------------
# N-QUEENS PROBLEM USING BACKTRACKING
# ---------------------------------------------------------

def is_safe(board, row, col, n):
    """
    Check whether a queen can be placed at (row, col).
    """

    # Check the same column
    for i in range(row):
        if board[i] == col:
            return False

        # Check diagonal
        if abs(board[i] - col) == abs(i - row):
            return False

    return True


# ---------------------------------------------------------
# BACKTRACKING FUNCTION
# ---------------------------------------------------------

def solve_n_queens(n):
    """
    Find all solutions for N-Queens.

    board[row] = column where the queen is placed.
    """

    board = [-1] * n
    solutions = []

    def backtrack(row):

        # All queens have been placed
        if row == n:
            solutions.append(board.copy())
            return

        # Try placing queen in every column
        for col in range(n):

            if is_safe(board, row, col, n):

                # Place queen
                board[row] = col

                # Move to next row
                backtrack(row + 1)

                # Remove queen (backtracking)
                board[row] = -1

    backtrack(0)

    return solutions


# ---------------------------------------------------------
# DISPLAY CHESSBOARD
# ---------------------------------------------------------

def print_board(solution, n):

    for row in range(n):

        for col in range(n):

            if solution[row] == col:
                print("Q", end=" ")
            else:
                print(".", end=" ")

        print()

    print()


# ---------------------------------------------------------
# MAIN PROGRAM
# ---------------------------------------------------------

def main():

    # -----------------------------------------------------
    # N = 4
    # -----------------------------------------------------

    n = 4

    print("=" * 50)
    print("N-QUEENS PROBLEM - N = 4")
    print("=" * 50)

    solutions_4 = solve_n_queens(n)

    print(f"\nTotal solutions for N = 4: {len(solutions_4)}\n")

    print("All solutions:")
    print("-" * 50)

    for i, solution in enumerate(solutions_4, start=1):

        print(f"Solution {i}:")
        print_board(solution, n)

    # -----------------------------------------------------
    # N = 8
    # -----------------------------------------------------

    n = 8

    print("=" * 50)
    print("N-QUEENS PROBLEM - N = 8")
    print("=" * 50)

    solutions_8 = solve_n_queens(n)

    print(f"\nTotal solutions for N = 8: {len(solutions_8)}")


# ---------------------------------------------------------
# EXECUTE PROGRAM
# ---------------------------------------------------------

if __name__ == "__main__":
    main()
```

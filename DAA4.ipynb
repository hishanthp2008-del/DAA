```python
# ---------------------------------------------------------
# SINGLE SOURCE SHORTEST PATH
# Dijkstra's Algorithm
#
# Vertices: 0, 1, 2, 3, 4, 5
# Source Vertex: 0
# ---------------------------------------------------------


# ---------------------------------------------------------
# GRAPH REPRESENTATION
# ---------------------------------------------------------

# Directed weighted graph
graph = {
    0: [(1, 4), (2, 2)],
    1: [(2, 1), (3, 5)],
    2: [(1, 1), (3, 8), (4, 10)],
    3: [(4, 2), (5, 6)],
    4: [(5, 3)],
    5: []
}


# ---------------------------------------------------------
# DIJKSTRA'S ALGORITHM
# ---------------------------------------------------------

def dijkstra(graph, source):

    # Number of vertices
    vertices = len(graph)

    # Initialize distances as infinity
    distance = [float('inf')] * vertices

    # Store previous vertex for path reconstruction
    previous = [None] * vertices

    # Source distance is 0
    distance[source] = 0

    # Set of unvisited vertices
    unvisited = set(range(vertices))

    while unvisited:

        # Find the unvisited vertex with minimum distance
        current = None

        for vertex in unvisited:
            if current is None or distance[vertex] < distance[current]:
                current = vertex

        # If remaining vertices are unreachable
        if distance[current] == float('inf'):
            break

        # Remove current vertex from unvisited set
        unvisited.remove(current)

        # Relax all outgoing edges
        for neighbor, weight in graph[current]:

            new_distance = distance[current] + weight

            if new_distance < distance[neighbor]:
                distance[neighbor] = new_distance
                previous[neighbor] = current

    return distance, previous


# ---------------------------------------------------------
# RECONSTRUCT SHORTEST PATH
# ---------------------------------------------------------

def get_path(previous, source, destination):

    path = []
    current = destination

    while current is not None:
        path.append(current)

        if current == source:
            break

        current = previous[current]

    # Destination is unreachable
    if path[-1] != source:
        return []

    # Reverse path
    path.reverse()

    return path


# ---------------------------------------------------------
# MAIN PROGRAM
# ---------------------------------------------------------

def main():

    source = 0

    # Run Dijkstra's algorithm
    distance, previous = dijkstra(graph, source)

    print("=" * 60)
    print("SINGLE SOURCE SHORTEST PATH - DIJKSTRA'S ALGORITHM")
    print("=" * 60)

    print("\nSource Vertex:", source)

    print("\nGraph Edges:")
    for vertex in graph:
        for neighbor, weight in graph[vertex]:
            print(f"{vertex} -> {neighbor}  (weight = {weight})")

    print("\n" + "-" * 60)
    print(f"{'Destination':<15}{'Distance':<15}{'Shortest Path'}")
    print("-" * 60)

    # Print shortest distance and path to every vertex
    for destination in range(len(graph)):

        path = get_path(
            previous,
            source,
            destination
        )

        if distance[destination] == float('inf'):
            print(
                f"{destination:<15}"
                f"{'INF':<15}"
                f"No path"
            )
        else:
            path_string = " -> ".join(
                map(str, path)
            )

            print(
                f"{destination:<15}"
                f"{distance[destination]:<15}"
                f"{path_string}"
            )

    print("-" * 60)


# ---------------------------------------------------------
# EXECUTE PROGRAM
# ---------------------------------------------------------

if __name__ == "__main__":
    main()
```

```python
# ---------------------------------------------------------
# MATRIX CHAIN MULTIPLICATION USING DYNAMIC PROGRAMMING
# ---------------------------------------------------------

# Matrices:
# A1 = 10 x 30
# A2 = 30 x 5
# A3 = 5 x 60
# A4 = 60 x 10
#
# Dimension array:
# p = [10, 30, 5, 60, 10]


# ---------------------------------------------------------
# FUNCTION TO FIND OPTIMAL MATRIX CHAIN ORDER
# ---------------------------------------------------------

def matrix_chain_order(p):

    n = len(p) - 1

    # m[i][j] stores minimum scalar multiplications
    # needed to multiply Ai...Aj
    m = [[0 for _ in range(n + 1)] for _ in range(n + 1)]

    # s[i][j] stores the position where the optimal
    # split occurs
    s = [[0 for _ in range(n + 1)] for _ in range(n + 1)]

    # Chain length
    for length in range(2, n + 1):

        for i in range(1, n - length + 2):

            j = i + length - 1

            # Initialize with infinity
            m[i][j] = float('inf')

            # Try every possible split
            for k in range(i, j):

                cost = (
                    m[i][k]
                    + m[k + 1][j]
                    + p[i - 1] * p[k] * p[j]
                )

                # Update minimum cost
                if cost < m[i][j]:
                    m[i][j] = cost
                    s[i][j] = k

    return m, s


# ---------------------------------------------------------
# FUNCTION TO PRINT OPTIMAL PARENTHESIZATION
# ---------------------------------------------------------

def print_optimal_parenthesis(s, i, j):

    if i == j:
        return f"A{i}"

    k = s[i][j]

    left = print_optimal_parenthesis(s, i, k)
    right = print_optimal_parenthesis(s, k + 1, j)

    return f"({left} x {right})"


# ---------------------------------------------------------
# MAIN PROGRAM
# ---------------------------------------------------------

def main():

    # Matrix dimensions
    p = [10, 30, 5, 60, 10]

    matrix_names = [
        "A1 (10x30)",
        "A2 (30x5)",
        "A3 (5x60)",
        "A4 (60x10)"
    ]

    print("=" * 65)
    print("MATRIX CHAIN MULTIPLICATION USING DYNAMIC PROGRAMMING")
    print("=" * 65)

    print("\nMatrices:")
    for matrix in matrix_names:
        print(matrix)

    # Calculate optimal order
    m, s = matrix_chain_order(p)

    # Generate optimal parenthesization
    optimal_order = print_optimal_parenthesis(
        s, 1, len(p) - 1
    )

    print("\n" + "-" * 65)

    print("Optimal Parenthesization:")
    print(optimal_order)

    print("\nMinimum Scalar Multiplications:")
    print(m[1][len(p) - 1])

    # -----------------------------------------------------
    # DISPLAY DP TABLE
    # -----------------------------------------------------

    print("\n" + "-" * 65)
    print("DP COST TABLE")
    print("-" * 65)

    print("       A1       A2       A3       A4")

    for i in range(1, 5):
        print(f"A{i}", end="  ")

        for j in range(1, 5):

            if i <= j:
                print(f"{m[i][j]:>7}", end="   ")
            else:
                print("       -", end="   ")

        print()

    # -----------------------------------------------------
    # DISPLAY SPLIT TABLE
    # -----------------------------------------------------

    print("\n" + "-" * 65)
    print("SPLIT TABLE")
    print("-" * 65)

    print("       A1       A2       A3       A4")

    for i in range(1, 5):
        print(f"A{i}", end="  ")

        for j in range(1, 5):

            if i < j:
                print(f"{s[i][j]:>7}", end="   ")
            else:
                print("       -", end="   ")

        print()


# ---------------------------------------------------------
# EXECUTE PROGRAM
# ---------------------------------------------------------

if __name__ == "__main__":
    main()
```

```python
# ---------------------------------------------------------
# EFFICIENT BIN PACKING USING APPROXIMATION ALGORITHMS
# ---------------------------------------------------------
# Items:
# [0.5, 0.7, 0.3, 0.9, 0.2, 0.6, 0.8, 0.4, 0.1, 0.5]
#
# Bin capacity = 1.0
#
# Algorithms:
# 1. Next Fit
# 2. First Fit
# 3. Best Fit
# ---------------------------------------------------------

import math


# ---------------------------------------------------------
# INPUT
# ---------------------------------------------------------

items = [0.5, 0.7, 0.3, 0.9, 0.2,
         0.6, 0.8, 0.4, 0.1, 0.5]

CAPACITY = 1.0


# ---------------------------------------------------------
# THEORETICAL LOWER BOUND
# ---------------------------------------------------------

def lower_bound(items, capacity):
    """
    The theoretical lower bound is:

        ceil(total weight / bin capacity)

    No packing can use fewer bins than this.
    """

    total_weight = sum(items)

    return math.ceil(total_weight / capacity)


# ---------------------------------------------------------
# 1. NEXT FIT
# ---------------------------------------------------------

def next_fit(items, capacity):
    """
    Next Fit:
    - Keep only the current bin open.
    - If the next item does not fit, close the current bin
      and open a new bin.
    """

    bins = []

    current_bin = []
    current_weight = 0.0

    for item in items:

        if current_weight + item <= capacity + 1e-9:

            current_bin.append(item)
            current_weight += item

        else:

            bins.append(current_bin)

            current_bin = [item]
            current_weight = item

    # Add the last bin
    if current_bin:
        bins.append(current_bin)

    return bins


# ---------------------------------------------------------
# 2. FIRST FIT
# ---------------------------------------------------------

def first_fit(items, capacity):
    """
    First Fit:
    Place each item into the first bin in which it fits.
    If no bin can accommodate it, create a new bin.
    """

    bins = []
    bin_weights = []

    for item in items:

        placed = False

        for i in range(len(bins)):

            if bin_weights[i] + item <= capacity + 1e-9:

                bins[i].append(item)
                bin_weights[i] += item

                placed = True
                break

        # Create a new bin if item does not fit
        if not placed:

            bins.append([item])
            bin_weights.append(item)

    return bins


# ---------------------------------------------------------
# 3. BEST FIT
# ---------------------------------------------------------

def best_fit(items, capacity):
    """
    Best Fit:
    Place the item into the bin that will have the
    smallest remaining space after insertion.
    """

    bins = []
    bin_weights = []

    for item in items:

        best_index = -1
        best_remaining = float('inf')

        for i in range(len(bins)):

            remaining = capacity - (
                bin_weights[i] + item
            )

            if remaining >= -1e-9:

                if remaining < best_remaining:

                    best_remaining = remaining
                    best_index = i

        # If a suitable bin was found
        if best_index != -1:

            bins[best_index].append(item)
            bin_weights[best_index] += item

        # Otherwise create a new bin
        else:

            bins.append([item])
            bin_weights.append(item)

    return bins


# ---------------------------------------------------------
# PRINT BIN DETAILS
# ---------------------------------------------------------

def print_bins(name, bins, capacity):

    print("\n" + "=" * 65)
    print(name)
    print("=" * 65)

    for i, bin_items in enumerate(bins, start=1):

        weight = sum(bin_items)
        remaining = capacity - weight

        print(
            f"Bin {i}: "
            f"{bin_items} "
            f"-> Weight = {weight:.1f}, "
            f"Remaining = {remaining:.1f}"
        )

    print("-" * 65)
    print(f"Number of bins used: {len(bins)}")


# ---------------------------------------------------------
# MAIN PROGRAM
# ---------------------------------------------------------

def main():

    print("=" * 65)
    print("BIN PACKING USING APPROXIMATION ALGORITHMS")
    print("=" * 65)

    print("\nItems:")
    print(items)

    print(f"Bin capacity: {CAPACITY}")

    total_weight = sum(items)

    print(f"Total item weight: {total_weight:.1f}")

    # Calculate theoretical lower bound
    lb = lower_bound(items, CAPACITY)

    print(f"Theoretical lower bound: {lb} bins")

    # -----------------------------------------------------
    # Run algorithms
    # -----------------------------------------------------

    nf_bins = next_fit(items, CAPACITY)
    ff_bins = first_fit(items, CAPACITY)
    bf_bins = best_fit(items, CAPACITY)

    # -----------------------------------------------------
    # Display individual results
    # -----------------------------------------------------

    print_bins(
        "1. NEXT FIT (NF)",
        nf_bins,
        CAPACITY
    )

    print_bins(
        "2. FIRST FIT (FF)",
        ff_bins,
        CAPACITY
    )

    print_bins(
        "3. BEST FIT (BF)",
        bf_bins,
        CAPACITY
    )

    # -----------------------------------------------------
    # Comparison
    # -----------------------------------------------------

    print("\n\n" + "=" * 65)
    print("COMPARISON OF APPROXIMATION ALGORITHMS")
    print("=" * 65)

    print(
        f"{'Algorithm':<20}"
        f"{'Bins Used':<15}"
        f"{'Lower Bound':<15}"
        f"{'Extra Bins'}"
    )

    print("-" * 65)

    results = [
        ("Next Fit", len(nf_bins)),
        ("First Fit", len(ff_bins)),
        ("Best Fit", len(bf_bins))
    ]

    for name, bins_used in results:

        extra = bins_used - lb

        print(
            f"{name:<20}"
            f"{bins_used:<15}"
            f"{lb:<15}"
            f"{extra}"
        )

    print("-" * 65)

    # -----------------------------------------------------
    # Find best algorithm
    # -----------------------------------------------------

    minimum_bins = min(
        len(nf_bins),
        len(ff_bins),
        len(bf_bins)
    )

    print(
        f"\nTheoretical lower bound = {lb} bins"
    )

    print(
        f"Best result = {minimum_bins} bins"
    )

    if minimum_bins == lb:
        print(
            "✓ At least one approximation algorithm "
            "achieved the theoretical lower bound."
        )
    else:
        print(
            "No algorithm achieved the theoretical "
            "lower bound for this ordering."
        )


# ---------------------------------------------------------
# EXECUTE PROGRAM
# ---------------------------------------------------------

if __name__ == "__main__":
    main()
```

```python
# ---------------------------------------------------------
# FIND MINIMUM AND MAXIMUM USING DIVIDE AND CONQUER
# ---------------------------------------------------------

import random


# ---------------------------------------------------------
# 1. NAIVE APPROACH
# ---------------------------------------------------------

def naive_min_max(arr):
    """
    Find minimum and maximum using a simple linear scan.
    Count the number of comparisons.
    """

    comparisons = 0

    min_value = arr[0]
    max_value = arr[0]

    for i in range(1, len(arr)):

        # Compare for minimum
        comparisons += 1
        if arr[i] < min_value:
            min_value = arr[i]

        # Compare for maximum
        comparisons += 1
        if arr[i] > max_value:
            max_value = arr[i]

    return min_value, max_value, comparisons


# ---------------------------------------------------------
# 2. DIVIDE AND CONQUER APPROACH
# ---------------------------------------------------------

def divide_conquer_min_max(arr, low, high):
    """
    Find minimum and maximum using Divide and Conquer.
    Returns:
        minimum
        maximum
        number of comparisons
    """

    # Base case: only one element
    if low == high:
        return arr[low], arr[low], 0

    # Base case: two elements
    if high == low + 1:

        comparisons = 1

        if arr[low] < arr[high]:
            return arr[low], arr[high], comparisons
        else:
            return arr[high], arr[low], comparisons

    # Find middle
    mid = (low + high) // 2

    # Divide
    left_min, left_max, left_comparisons = divide_conquer_min_max(
        arr, low, mid
    )

    right_min, right_max, right_comparisons = divide_conquer_min_max(
        arr, mid + 1, high
    )

    # Combine
    comparisons = left_comparisons + right_comparisons

    # Compare minimum values
    comparisons += 1
    if left_min < right_min:
        minimum = left_min
    else:
        minimum = right_min

    # Compare maximum values
    comparisons += 1
    if left_max > right_max:
        maximum = left_max
    else:
        maximum = right_max

    return minimum, maximum, comparisons


# ---------------------------------------------------------
# MAIN PROGRAM
# ---------------------------------------------------------

def main():

    # Required array sizes
    sizes = [10, 100, 1000, 10000]

    print("=" * 75)
    print("MIN-MAX USING DIVIDE AND CONQUER")
    print("=" * 75)

    print(
        f"{'Array Size':<15}"
        f"{'Naive Comparisons':<22}"
        f"{'D&C Comparisons':<22}"
        f"{'Min / Max'}"
    )

    print("-" * 75)

    for size in sizes:

        # Generate random array
        arr = [
            random.randint(1, 100000)
            for _ in range(size)
        ]

        # Naive approach
        naive_min, naive_max, naive_comparisons = naive_min_max(arr)

        # Divide and Conquer approach
        dc_min, dc_max, dc_comparisons = divide_conquer_min_max(
            arr, 0, size - 1
        )

        # Verify both methods produce the same result
        if naive_min != dc_min or naive_max != dc_max:
            print("Error: Results do not match!")

        print(
            f"{size:<15}"
            f"{naive_comparisons:<22}"
            f"{dc_comparisons:<22}"
            f"{dc_min} / {dc_max}"
        )

    print("-" * 75)


# ---------------------------------------------------------
# EXECUTE PROGRAM
# ---------------------------------------------------------

if __name__ == "__main__":
    main()
```
